```python
import numpy as np

# Generate random marks (50 students, 5 subjects)
marks = np.random.randint(35, 100, size=(50, 5))

print("Marks Matrix:\n", marks)

# Average per student
student_avg = np.mean(marks, axis=1)

# Average per subject
subject_avg = np.mean(marks, axis=0)

print("\nAverage per Student:\n", student_avg)
print("\nAverage per Subject:\n", subject_avg)

# Highest and lowest marks
print("\nHighest Mark:", np.max(marks))
print("Lowest Mark:", np.min(marks))

# Top 5 students
top5 = np.argsort(student_avg)[-5:]
print("\nTop 5 Students Index:", top5)

# Pass count (average > 40)
pass_count = np.sum(student_avg > 40)
print("\nPass Count:", pass_count)
```

```python
import numpy as np

def is_valid_group(group):
    """Check if group contains numbers 1-9 exactly once"""
    return np.array_equal(np.sort(group), np.arange(1, 10))

def validate_sudoku(board):
    board = np.array(board)

    # Check shape
    if board.shape != (9, 9):
        return False

    # Check rows
    for row in board:
        if not is_valid_group(row):
            return False

    # Check columns
    for col in board.T:
        if not is_valid_group(col):
            return False

    # Check 3x3 subgrids
    for i in range(0, 9, 3):
        for j in range(0, 9, 3):
            subgrid = board[i:i+3, j:j+3].flatten()
            if not is_valid_group(subgrid):
                return False

    return True
```
